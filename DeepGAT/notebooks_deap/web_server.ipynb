```python
import http.server
import socketserver
import os

# Portul pe care va rula interfața web (ex: http://localhost:8000)
PORT = 8000

# Directorul curent (presupunem că index.html este în același folder cu acest script)
DIRECTORY = "."

class Handler(http.server.SimpleHTTPRequestHandler):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, directory=DIRECTORY, **kwargs)

    # Oprim log-urile excesive în consolă pentru fiecare cerere de imagine/css
    def log_message(self, format, *args):
        pass

print(f"🌐 Serverul Web a pornit!")
print(f"👉 Deschide browser-ul și accesează: http://localhost:{PORT}")
print("Apasa Ctrl+C pentru a opri.")

try:
    with socketserver.TCPServer(("", PORT), Handler) as httpd:
        httpd.serve_forever()
except KeyboardInterrupt:
    print("\nServer Web oprit.")
```

```python
import asyncio
import websockets
import json
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import pyeeg as pe
import warnings
import os
import nest_asyncio
from collections import deque
import threading
import http.server
import socketserver
from sklearn.preprocessing import StandardScaler

nest_asyncio.apply()
warnings.filterwarnings('ignore')

# ─────────────────────────────────────────────────────────────────────────────
# 1. CONFIGURARE SERVER (PC)
# ─────────────────────────────────────────────────────────────────────────────
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BASE_PATH = r"C:\Users\PC\Desktop\Facultate\LICENTA\notebooks\DeepGAT\gat_output_robust"
MODEL_PATH = os.path.join(BASE_PATH, 'best_model.pth')

LOCAL_PORT = 65432       # Portul WebSocket pentru comunicarea cu laptopul
HTTP_PORT = 8000         # Portul pentru dashboard-ul Web
WINDOW_SIZE = 256        # 2 secunde la 128Hz
INFERENCE_STEP = 32      # Rulăm inferența la fiecare 32 de sample-uri noi
BUFFER_CAPACITY = int(5 / (INFERENCE_STEP / 128)) 

N_CH = 16
N_FEATS = 10

# ─────────────────────────────────────────────────────────────────────────────
# 2. DEFINIȚII MODEL GAT
# ─────────────────────────────────────────────────────────────────────────────
class GATLayer(nn.Module):
    def __init__(self, in_features: int, out_features: int, num_heads: int = 4, attn_dropout: float = 0.0, residual: bool = True):
        super().__init__()
        self.H = num_heads
        self.d = out_features
        self.residual = residual
        self.W = nn.Linear(in_features, num_heads * out_features, bias=False)
        self.a = nn.Parameter(torch.empty(num_heads, 2 * out_features))
        nn.init.xavier_uniform_(self.a.unsqueeze(0))
        self.leaky = nn.LeakyReLU(0.2)
        self.attn_drop = nn.Dropout(attn_dropout)
        self.bn = nn.BatchNorm1d(out_features)
        if residual:
            self.res_proj = (nn.Linear(in_features, out_features, bias=False) 
                             if in_features != out_features else nn.Identity())

    def forward(self, x: torch.Tensor):
        B, N, _ = x.shape
        h = self.W(x).view(B, N, self.H, self.d)
        hi = h.unsqueeze(2)
        hj = h.unsqueeze(1)
        pair = torch.cat([hi.expand(B, N, N, self.H, self.d),
                          hj.expand(B, N, N, self.H, self.d)], dim=-1)
        e = self.leaky((pair * self.a.unsqueeze(0).unsqueeze(0).unsqueeze(0)).sum(-1))
        alpha = self.attn_drop(F.softmax(e, dim=2))            
        out = torch.einsum("bqkh, bkhd -> bqhd", alpha, h).mean(dim=2)
        out = self.bn(out.reshape(B * N, self.d)).reshape(B, N, self.d)
        out = F.elu(out)
        if self.residual:
            out = out + self.res_proj(x)
        return out, alpha.permute(0, 3, 1, 2)

class TaskHead(nn.Module):
    def __init__(self, in_dim: int, dense: int, head_dropout: float = 0.0):
        super().__init__()
        self.head = nn.Sequential(nn.Linear(in_dim, dense), nn.LayerNorm(dense), nn.GELU(), nn.Dropout(head_dropout), nn.Linear(dense, 1))
    def forward(self, x): return self.head(x.mean(dim=1))

class DeepGAT(nn.Module):
    def __init__(self, n_channels: int, in_feats: int, backbone_dims: list, dense_size: int, num_heads: int = 4):
        super().__init__()
        d0 = backbone_dims[0]
        self.input_proj = nn.Linear(in_feats, d0)
        self.ch_embed = nn.Parameter(torch.randn(1, n_channels, d0) * 0.02)
        dims = [d0] + backbone_dims
        self.backbone = nn.ModuleList([GATLayer(dims[i], dims[i+1], num_heads, residual=True) for i in range(len(backbone_dims))])
        self.head_aro = TaskHead(backbone_dims[-1], dense_size)
        self.head_val = TaskHead(backbone_dims[-1], dense_size)

    def forward(self, x: torch.Tensor):
        x = F.gelu(self.input_proj(x)) + self.ch_embed            
        for layer in self.backbone: x, _ = layer(x)
        return torch.cat([self.head_aro(x), self.head_val(x)], dim=1)

# ─────────────────────────────────────────────────────────────────────────────
# 3. INIȚIALIZARE MODEL 
# ─────────────────────────────────────────────────────────────────────────────
print(f"Se inițializează GAT pe {DEVICE}...")
model = DeepGAT(n_channels=N_CH, in_feats=N_FEATS, backbone_dims=[64, 64, 64], dense_size=128, num_heads=4).to(DEVICE)

if os.path.exists(MODEL_PATH):
    model.load_state_dict(torch.load(MODEL_PATH, map_location=DEVICE))
    model.eval()
    print("✓ Model Deep GAT încărcat.")
else:
    print("⚠️ AVERTISMENT: Modelul nu a fost găsit. Ai setat calea corectă?")

# Scaler-ul nu se mai citește de pe disc, îl vom crea dinamic
print("✓ Mod de Standardizare Dinamică (Rolling Scaler) activat.")

# ─────────────────────────────────────────────────────────────────────────────
# 4. FUNCȚII EXTRAGERE TRĂSĂTURI (FEATURE EXTRACTION & DYNAMIC SCALING)
# ─────────────────────────────────────────────────────────────────────────────
# Istoricul trăsăturilor pentru a calcula o medie (aproximativ 30 de secunde de istoric)
feature_history = deque(maxlen=120)

def differential_entropy(sig, band_edges, fs):
    de, fft_full = [], np.fft.rfft(sig)
    freqs = np.fft.rfftfreq(len(sig), 1.0 / fs)
    for lo, hi in zip(band_edges[:-1], band_edges[1:]):
        mask = (freqs >= lo) & (freqs < hi)
        f = np.zeros_like(fft_full); f[mask] = fft_full[mask]
        var = np.var(np.fft.irfft(f, n=len(sig))) + 1e-10
        de.append(0.5 * np.log(2 * np.pi * np.e * var))
    return de

def extract_features(raw_window_data, sample_rate=128):
    band_edges = [4, 8, 12, 16, 25, 45] 
    feats = []
    for ch in range(raw_window_data.shape[0]):
        sig = raw_window_data[ch, :] 
        bp = list(pe.bin_power(sig, band_edges, sample_rate)[0])
        de = differential_entropy(sig, band_edges, sample_rate)
        feats.extend(bp + de)
    return np.array(feats, dtype=np.float32)

def preprocess_and_predict(features_vector):
    # features_vector este 1D cu 160 de elemente. Îl aducem la formatul de antrenament: (16, 10)
    feat_2d = features_vector.reshape(N_CH, N_FEATS)
    feature_history.append(feat_2d)
    
    # Transformăm istoricul într-un array (N_ferestre_istoric, 16, 10)
    history_array = np.array(feature_history)
    N_windows = history_array.shape[0]
    
    if N_windows > 1:
        scaler = StandardScaler()
        # Aplatizăm temporar pentru a potrivi structura din timpul antrenamentului
        scaled_flat = scaler.fit_transform(history_array.reshape(-1, N_FEATS))
        # Luăm doar ultima fereastră curentă (ultimele 16 canale normalizate)
        current_features = scaled_flat[-N_CH:].reshape(N_CH, N_FEATS)
    else:
        # Pentru chiar primul cadru primit, fallback cu zero-uri 
        current_features = np.zeros((N_CH, N_FEATS))
        
    return torch.tensor(current_features).float().unsqueeze(0).to(DEVICE)

# ─────────────────────────────────────────────────────────────────────────────
# 5. SERVER WEBSOCKET (Inference Pipeline)
# ─────────────────────────────────────────────────────────────────────────────
prediction_history = deque(maxlen=BUFFER_CAPACITY)
connected_clients = set()

async def broadcast_msg(msg_dict):
    if not connected_clients: return
    msg_str = json.dumps(msg_dict)
    tasks = [asyncio.create_task(c.send(msg_str)) for c in connected_clients]
    if tasks: await asyncio.wait(tasks)

async def handler(websocket):
    print("-> Client/Streamer conectat la WebSocket!")
    connected_clients.add(websocket)
    data_buffer = np.zeros((16, 0))
    samples_accumulated = 0  
    
    try:
        async for message in websocket:
            try:
                received_data = json.loads(message)

                # Rutare comenzi UI / Control
                if isinstance(received_data, dict):
                    msg_type = received_data.get("type")
                    if msg_type in ["cmd_start_stream", "cmd_pause_stream", "cmd_resume_stream", "stream_info", "stream_end", "progress"]:
                        cmd_msg = json.dumps(received_data)
                        tasks = [asyncio.create_task(c.send(cmd_msg)) for c in connected_clients if c != websocket]
                        if tasks: await asyncio.wait(tasks)
                        
                        if msg_type == "cmd_start_stream":
                            data_buffer = np.zeros((16, 0))
                            prediction_history.clear()
                            feature_history.clear() # Golește istoricul de normalizare la un stream nou
                        continue

                # Procesare Semnal EEG transmis de Laptop
                if isinstance(received_data, list):
                    chunk = np.array(received_data)
                    if chunk.shape[1] > 0:
                        # Trimite semnalul curat la UI
                        await broadcast_msg({"type": "signal", "data": chunk.tolist()})

                        # Adaugă în bufferul local
                        data_buffer = np.concatenate((data_buffer, chunk), axis=1)
                        samples_accumulated += chunk.shape[1] 
                        
                        # Păstrează doar fereastra maximă (256 samples)
                        if data_buffer.shape[1] > WINDOW_SIZE:
                            data_buffer = data_buffer[:, -WINDOW_SIZE:]
                        
                        # Când s-au acumulat noi `INFERENCE_STEP` samples și avem o fereastră plină -> INFERENȚĂ
                        if data_buffer.shape[1] == WINDOW_SIZE and samples_accumulated >= INFERENCE_STEP:
                            samples_accumulated = 0 
                            feats = extract_features(data_buffer, sample_rate=128)
                            tens = preprocess_and_predict(feats)
                            
                            with torch.no_grad():
                                out = model(tens)
                                probs = torch.sigmoid(out).cpu().numpy()[0]
                            
                            prediction_history.append((probs[0], probs[1]))
                            avg_aro = sum(p[0] for p in prediction_history) / len(prediction_history)
                            avg_val = sum(p[1] for p in prediction_history) / len(prediction_history)
                            
                            # Trimite predicția la UI
                            await broadcast_msg({"type": "prediction", "arousal": avg_aro, "valence": avg_val})

            except Exception as e: 
                pass
    except websockets.exceptions.ConnectionClosed:
        print("\n<- Client/Streamer deconectat.")
    finally:
        connected_clients.remove(websocket)

# ─────────────────────────────────────────────────────────────────────────────
# 6. HTTP SERVER PENTRU DASHBOARD (Background Thread)
# ─────────────────────────────────────────────────────────────────────────────
class QuietHandler(http.server.SimpleHTTPRequestHandler):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, directory=".", **kwargs)
    def log_message(self, format, *args):
        pass # Oprește spam-ul în consolă

def run_http_server():
    with socketserver.TCPServer(("", HTTP_PORT), QuietHandler) as httpd:
        print(f"🌐 Server HTTP pornit! Deschide: http://localhost:{HTTP_PORT}")
        httpd.serve_forever()

async def main():
    # Pornim serverul HTTP într-un thread separat ca să nu blocheze asyncio
    threading.Thread(target=run_http_server, daemon=True).start()
    
    print(f"📡 Server WebSocket pornit pe portul {LOCAL_PORT}...")
    async with websockets.serve(handler, "0.0.0.0", LOCAL_PORT, ping_interval=None):
        await asyncio.Future()

if __name__ == "__main__":
    try: 
        asyncio.run(main())
    except KeyboardInterrupt: 
        print("\nOprit.")
```

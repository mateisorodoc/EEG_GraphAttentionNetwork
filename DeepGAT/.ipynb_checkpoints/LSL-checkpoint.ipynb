```python
import time
import threading
import os
import pandas as pd
import requests
import io

from pylsl import StreamInlet, resolve_byprop

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

from webdriver_manager.chrome import ChromeDriverManager


SHEET_URL = "https://docs.google.com/spreadsheets/d/1xvwLRa_QE4QsopQWWJBoMO-lzheScZMt5OZLdEFLDng/export?format=csv&gid=0"
OUT_FOLDER = "recordings_happy"


def parse_time_to_seconds(time_val):
    if pd.isna(time_val):
        return 0

    time_str = str(time_val).strip()

    if ":" in time_str:
        try:
            m, s = map(int, time_str.split(":"))
            return m * 60 + s
        except:
            return 0

    try:
        return int(float(time_str))
    except:
        return 0


class LSLRecorder:
    def __init__(self):
        self.recording = False
        self.data = []
        self.inlet = None
        self.channel_labels = []
        self.time_offset = 0

    def connect(self):
        print("Căutare stream obci_eeg1...")
        streams = resolve_byprop('name', 'obci_eeg1', timeout=10)

        if not streams:
            raise Exception("LSL stream not found!")

        self.inlet = StreamInlet(streams[0])
        self.time_offset = self.inlet.time_correction()

        count = self.inlet.channel_count if isinstance(self.inlet.channel_count, int) else self.inlet.channel_count()
        self.channel_labels = [f"EEG_{i+1}" for i in range(count)]

        print("✓ LSL connected")
        print("Channels:", self.channel_labels)

    def record_loop(self, video_name, start_time):
        while self.recording:
            sample, ts = self.inlet.pull_sample(timeout=0.1)

            if sample:
                corrected_ts = ts + self.time_offset
                relative_time = corrected_ts - start_time

                self.data.append(
                    [corrected_ts, relative_time] + sample + [video_name]
                )

    def wait_for_video_and_play(self, driver):
        wait = WebDriverWait(driver, 20)
        wait.until(EC.presence_of_element_located((By.TAG_NAME, "video")))

        for _ in range(10):
            try:
                video = driver.find_element(By.TAG_NAME, "video")

                driver.execute_script("arguments[0].play();", video)

                is_playing = driver.execute_script(
                    "return arguments[0].paused === false;", video
                )

                if is_playing:
                    return True

            except:
                pass

            time.sleep(1)

        raise Exception("Video did not start")

    def run(self):
        # Load sheet
        headers = {'User-Agent': 'Mozilla/5.0'}
        response = requests.get(SHEET_URL, headers=headers)
        df = pd.read_csv(io.StringIO(response.text), encoding='utf-8')

        self.connect()

        # 🔥 Attach to existing Chrome
        opts = Options()
        opts.add_experimental_option("debuggerAddress", "localhost:9222")

        driver = webdriver.Chrome(
            service=Service(ChromeDriverManager().install()),
            options=opts
        )

        if not os.path.exists(OUT_FOLDER):
            os.makedirs(OUT_FOLDER)

        for i, row in df.iterrows():
            try:
                name = str(row.iloc[2])
                raw_url = str(row.iloc[4])
                start_s = parse_time_to_seconds(row.iloc[5])
                end_s = parse_time_to_seconds(row.iloc[6])

                # extract video ID
                if "v=" in raw_url:
                    v_id = raw_url.split("v=")[-1].split("&")[0]
                else:
                    v_id = raw_url.split("/")[-1].split("?")[0]

                url = f"https://www.youtube.com/watch?v={v_id}&t={start_s}s"

                print(f"\n▶️ {name} ({start_s}-{end_s}s)")

                self.data = []
                self.recording = True

                driver.get(url)
                self.wait_for_video_and_play(driver)

                # 🎯 LSL-aligned start
                start_time = self.inlet.local_clock()

                rec_thread = threading.Thread(
                    target=self.record_loop,
                    args=(name, start_time)
                )
                rec_thread.start()

                duration = max(0, end_s - start_s)
                time.sleep(duration)

                self.recording = False
                rec_thread.join()

                if self.data:
                    safe_name = "".join(
                        c for c in name if c.isalnum() or c in (' ', '_')
                    ).strip()

                    filename = f"{OUT_FOLDER}/EEG_{safe_name.replace(' ','_')}.csv"

                    columns = (
                        ["timestamp", "relative_time"]
                        + self.channel_labels
                        + ["video"]
                    )

                    pd.DataFrame(self.data, columns=columns).to_csv(filename, index=False)

                    print(f"✓ Saved: {filename}")

                time.sleep(2)

            except Exception as e:
                print(f"⚠️ Error: {e}")

        driver.quit()


if __name__ == "__main__":
    try:
        LSLRecorder().run()
    except Exception as e:
        print(f"Fatal error: {e}")
```
